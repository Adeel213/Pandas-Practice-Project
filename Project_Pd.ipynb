### Reading the CSV File

### Cleaning the Dataset

```python
new_sal = sal.drop(sal[sal['EmployeeName'] == 'Not provided'].index)
```

```python
new_sal = new_sal.drop(new_sal[new_sal['JobTitle'] == 'Not provided'].index)
```

```python
new_sal = new_sal.drop(new_sal[new_sal['BasePay'] == 'Not provided'].index)
```

```python
new_sal.info()
```

```python
new_sal[['BasePay','OvertimePay','OtherPay','Benefits']] = new_sal[['BasePay','OvertimePay','OtherPay','Benefits']].astype('float64')
```

```python
new_sal.isna().sum()
```

```python
new_sal['BasePay'] = new_sal['BasePay'].fillna(new_sal['BasePay'].mean())
```

```python
new_sal['Benefits'] = new_sal['Benefits'].fillna(new_sal['Benefits'].median())
```

```python
new_sal.drop(columns = ['Notes','Status'],inplace = True)
```

```python
new_sal
```

```python
sal = new_sal
```

```python
sal.info()
```
