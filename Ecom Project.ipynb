```python
import numpy as np
import pandas as pd
```

### Loading And Reading The Dataset

```python
ecom = pd.read_csv('Ecommerce Purchases.csv')
ecom.head()
```

```python
ecom.info()
```

### Average Purchase Price 

```python
print(ecom['Purchase Price'].mean())
```

### Highest and lowest Purchase Price

```python
print(f'Highest Price is: {ecom['Purchase Price'].max()}')
print(f'Lowest price is: {ecom['Purchase Price'].min()}')
```

### How many people have Language English

```python
print(ecom['Language'].str.contains('en').sum())
```

### How many people's Job is 'Laywer'

```python
print(ecom[ecom['Job'] == 'Lawyer']['Job'].count())
```

### How many people made purchase at AM and how many at PM

```python
ecom['AM or PM'].value_counts()
```

### Top 5 jobs

```python
ecom['Job'].value_counts().head()
```

### Someone made purchase that came from lot '90-WT', what was the purchase price for this transcation

```python
ecom[ecom['Lot'] == '90 WT']['Purchase Price']
```

### What is the Email of this Credit Card number owner '4926535242672853'

```python
ecom[ecom['Credit Card'] == 4926535242672853]['Email']
```

### How many people have American Express as their Credit Card and made a purchase above $95

```python
ecom[(ecom['CC Provider'] == 'American Express') & (ecom['Purchase Price']>95)].count()
```

### How many have Credit Card that Expire in 2025

```python
print(ecom['CC Exp Date'].apply(lambda x: x.split('/')[1] == '25').sum())
```

### What are the top 5 most popular Email provider

```python
ecom['Email'].apply(lambda x: x.split('@')[1]).value_counts().head()
```
